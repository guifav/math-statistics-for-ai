## 1. Vetores em Machine Learning <a id='1-vetores-ml'></a>

### Intuicao: O que e um Vetor?

Pense em um vetor como um **ponto em um espaco multidimensional** ou uma **seta saindo da origem**. Cada vez que voce descreve algo com multiplos numeros (altura, peso, idade), voce está criando um vetor.

**Analogia do Mundo Real:**
- Voce tem 3 amigos com as seguintes caracteristicas: idade (25, 32, 24) anos
- Os pesos de suas mochilas: 5, 8, 4 kg
- Essas listas de numeros sao vetores!

Em Machine Learning, **cada exemplo no seu dataset e um vetor**. Se voce tem uma foto de 28x28 pixels em preto-e-branco, ao "linearizar" essa matriz voce obtem um vetor com 784 numeros (um por pixel).

Em ML, **um vetor e uma observacao**. Cada exemplo no seu dataset e representado
como um vetor de features:

$$\mathbf{x} = [x_1, x_2, \ldots, x_d] \in \mathbb{R}^d$$

| Aplicacao | Vetor | Dimensao tipica |
|-----------|-------|-----------------|
| Imagem (grayscale 28x28) | Pixels linearizados | 784 |
| Texto (bag of words) | Contagem de palavras | 10.000-100.000 |
| Embedding de palavra | Representacao densa | 100-1536 |
| Tabular (Boston Housing) | Atributos da casa | 13 |
| Sinal de audio (1s, 44kHz) | Amostras de amplitude | 44.000 |

**Intuicao geometrica:** Um vetor e uma seta no espaco $d$-dimensional,
com direcao e magnitude. Dois exemplos "similares" terao vetores proximos.

### O que observar
- Vetores podem representar pontos, direções ou magnitudes
- Em Python, usamos arrays NumPy para implementar vetores
- Operações vetoriais são eficientes (broadcasting automático)

## 2. Definição e Representação <a id='2-definicao'></a>

Um vetor em $\mathbb{R}^n$ é uma lista ordenada de $n$ números reais:

$$\mathbf{v} = \begin{pmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{pmatrix} \text{ (coluna)} \quad \text{ou} \quad \mathbf{v}^\top = [v_1, v_2, \ldots, v_n] \text{ (linha)}$$

### Tipos de Vetores Especiais

| Nome | Definição | Uso em ML |
|------|-----------|-----------|
| **Vetor zero** | $\mathbf{0} = [0, 0, \ldots, 0]$ | Inicialização de bias |
| **Vetor unitário** | $\|\mathbf{e}_i\| = 1$ | Base canônica, one-hot encoding |
| **One-hot** | $[0, \ldots, 1, \ldots, 0]$ | Encoding de categorias |
| **Vetor de uns** | $\mathbf{1} = [1, 1, \ldots, 1]$ | Cálculo de média |


### O que concluir
- A dimensionalidade não limita os conceitos: funcionam em 2D, 3D, e n-dimensões
- Visualizações ajudam a intuição, mas cálculos funcionam sem visualizar

### O que concluir
Vetores são a unidade fundamental: tudo em ML (features, pesos, gradientes) são vetores.

## 3. Operacoes Basicas com Vetores <a id='3-operacoes-basicas'></a>

### Intuicao: Operacoes como Transformacoes em Dados

Antes de entrar nas formulas, pense assim: quando treinamos um modelo de ML, estamos constantemente manipulando vetores. Cada passo do gradient descent envolve tres operacoes vetoriais fundamentais:

1. **Subtracao:** calcular o erro (predicao - real)
2. **Multiplicacao por escalar:** ajustar pelo learning rate
3. **Adicao/Subtracao:** atualizar os pesos

$$\mathbf{w}_{\text{novo}} = \mathbf{w}_{\text{atual}} - \eta \cdot \nabla L$$

Onde $\eta$ e o learning rate (escalar) e $\nabla L$ e o gradiente (vetor).

### Definicoes Formais

| Operacao | Formula | Geometria |
|----------|---------|-----------|
| **Adicao** | $\mathbf{u}+\mathbf{v} = [u_1+v_1, \ldots, u_n+v_n]$ | Regra tip-to-tail |
| **Subtracao** | $\mathbf{u}-\mathbf{v} = [u_1-v_1, \ldots, u_n-v_n]$ | Vetor de v ate u |
| **Escalar** | $c\mathbf{v} = [cv_1, \ldots, cv_n]$ | Estica/encolhe/inverte |

### por que em ml
- O produto interno mede *semelhança* entre vetores (fundamental em regressão, clustering)
- O ângulo entre vetores é usado em similaridade de cosseno (busca textual, recomendação)

### por que em ml
Operações vetoriais (soma, multiplicação) são usadas em: atualização de pesos, combinação de features.


````python
# Visualizacao: operacoes vetoriais
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

u = np.array([2, 1])
v = np.array([1, 2])

# --- Painel 1: Adicao (tip-to-tail) ---
axes[0].set_xlim(-1, 5); axes[0].set_ylim(-1, 4)
axes[0].axhline(0, color='k', lw=0.3); axes[0].axvline(0, color='k', lw=0.3)
axes[0].set_aspect('equal')

# u (azul, da origem)
axes[0].annotate('', xy=u, xytext=(0,0), arrowprops=dict(arrowstyle='->', color='blue', lw=2))
# v (vermelho, da ponta de u -- tip-to-tail)
axes[0].annotate('', xy=u+v, xytext=u, arrowprops=dict(arrowstyle='->', color='red', lw=2))
# u+v (verde tracejado, da origem)
axes[0].annotate('', xy=u+v, xytext=(0,0), arrowprops=dict(arrowstyle='->', color='green', lw=2.5, ls='--'))
axes[0].text(1, 0.2, 'u', color='blue', fontsize=12, fontweight='bold')
axes[0].text(2.5, 2.2, 'v', color='red', fontsize=12, fontweight='bold')
axes[0].text(1.2, 1.8, 'u+v', color='green', fontsize=12, fontweight='bold')
axes[0].set_title('Adicao: regra tip-to-tail')

# --- Painel 2: Subtracao ---
axes[1].set_xlim(-2, 3); axes[1].set_ylim(-2, 3)
axes[1].axhline(0, color='k', lw=0.3); axes[1].axvline(0, color='k', lw=0.3)
axes[1].set_aspect('equal')

axes[1].annotate('', xy=u, xytext=(0,0), arrowprops=dict(arrowstyle='->', color='blue', lw=2))
axes[1].annotate('', xy=v, xytext=(0,0), arrowprops=dict(arrowstyle='->', color='red', lw=2))
axes[1].annotate('', xy=u-v, xytext=(0,0), arrowprops=dict(arrowstyle='->', color='purple', lw=2.5, ls='--'))
# Seta de v ate u para mostrar que u-v = deslocamento de v para u
axes[1].annotate('', xy=u, xytext=v, arrowprops=dict(arrowstyle='->', color='purple', lw=1.5, ls=':'))
axes[1].text(1.1, 0.2, 'u', color='blue', fontsize=12, fontweight='bold')
axes[1].text(0.3, 1.3, 'v', color='red', fontsize=12, fontweight='bold')
axes[1].text(0.7, -1.2, 'u-v', color='purple', fontsize=12, fontweight='bold')
axes[1].set_title('Subtracao: u - v')

# --- Painel 3: Multiplicacao por escalar ---
scalars = [-1, 0.5, 1, 2]
colors_s = ['red', 'orange', 'blue', 'green']
base = np.array([1.5, 1])
axes[2].set_xlim(-4, 4); axes[2].set_ylim(-2, 3); axes[2].set_aspect('equal')
axes[2].axhline(0, color='k', lw=0.3); axes[2].axvline(0, color='k', lw=0.3)
for s, c in zip(scalars, colors_s):
    scaled = s * base
    axes[2].annotate('', xy=scaled, xytext=(0,0),
                     arrowprops=dict(arrowstyle='->', color=c, lw=2))
    axes[2].text(scaled[0]+0.1, scaled[1]+0.1, f'{s}v', color=c, fontsize=11, fontweight='bold')
axes[2].set_title('Multiplicacao por escalar')

plt.tight_layout()
plt.show()

# --- Exemplo pratico: gradient update ---
print("=== Gradient Update (operacao vetorial na pratica) ===")
w = np.array([3.0, 1.5])
grad = np.array([0.8, -0.4])
lr = 0.5
w_new = w - lr * grad
print(f"Pesos atuais:   w = {w}")
print(f"Gradiente:      grad = {grad}")
print(f"Learning rate:  lr = {lr}")
print(f"Novo peso:      w_new = w - lr * grad = {w} - {lr}*{grad} = {w_new}")
print(f"\nO que aconteceu: subtracao vetorial (w - lr*grad) = multiplicacao por escalar + subtracao")
````

### O que observar nos graficos acima

**Adicao (esquerda):** Para somar u+v geometricamente, coloque a "cauda" de v na "ponta" de u. O vetor resultante (tracejado verde) vai da origem ate a ponta final. Essa regra tip-to-tail e exatamente o que acontece no gradient descent: o vetor de pesos "anda" somando (ou subtraindo) o gradiente.

**Subtracao (centro):** u-v e o vetor que vai da ponta de v ate a ponta de u. Em ML, a diferenca entre dois vetores mede a "distancia" entre eles -- e assim que KNN e K-Means funcionam.

**Escalar (direita):** Multiplicar por 2 dobra o comprimento. Multiplicar por 0.5 corta pela metade. Multiplicar por -1 inverte a direcao. No gradient descent, multiplicamos o gradiente pelo learning rate (um escalar pequeno, tipo 0.001) para dar passos controlados.
### conexao com
Operações com vetores se generalizam para matrizes: soma, multiplicação escalar, composição (0_3).


````python
# Visualizacao: produto interno e sua relacao com angulo
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# 1) Produto interno vs angulo
theta_range = np.linspace(0, np.pi, 200)
dot_values = np.cos(theta_range)  # para vetores unitarios, u.v = cos(theta)

axes[0].plot(np.degrees(theta_range), dot_values, 'b-', lw=2.5)
axes[0].fill_between(np.degrees(theta_range), dot_values, 0,
                     where=dot_values > 0, alpha=0.15, color='green', label='u.v > 0 (mesma direcao)')
axes[0].fill_between(np.degrees(theta_range), dot_values, 0,
                     where=dot_values < 0, alpha=0.15, color='red', label='u.v < 0 (opostos)')
axes[0].axhline(0, color='k', lw=0.8)
axes[0].axvline(90, color='gray', ls='--', lw=1, label='90 graus (ortogonal)')
axes[0].set_xlabel('Angulo entre vetores (graus)')
axes[0].set_ylabel('Produto interno (vetores unitarios)')
axes[0].set_title('Produto Interno vs Angulo')
axes[0].legend(fontsize=9)

# 2) Attention scores (simplificado)
np.random.seed(42)
n_tokens = 6
tokens = ['O', 'gato', 'sentou', 'no', 'tapete', '.']
Q = np.random.randn(n_tokens, 4)
K = np.random.randn(n_tokens, 4)
attention = Q @ K.T / np.sqrt(4)  # scaled dot-product

im = axes[1].imshow(attention, cmap='YlOrRd', aspect='auto')
axes[1].set_xticks(range(n_tokens)); axes[1].set_xticklabels(tokens, fontsize=10)
axes[1].set_yticks(range(n_tokens)); axes[1].set_yticklabels(tokens, fontsize=10)
axes[1].set_title('Attention Scores (Scaled Dot-Product)\nquery . key / sqrt(d)')
axes[1].set_xlabel('Keys'); axes[1].set_ylabel('Queries')
plt.colorbar(im, ax=axes[1], shrink=0.8)

plt.tight_layout()
plt.show()

# Demonstracao numerica
u = np.array([3.0, 1.0])
v = np.array([1.0, 2.0])
dot_uv = np.dot(u, v)
angle = np.degrees(np.arccos(dot_uv / (np.linalg.norm(u) * np.linalg.norm(v))))
print(f"u = {u}, v = {v}")
print(f"u . v = {dot_uv}  (positivo: apontam na mesma direcao geral)")
print(f"Angulo entre u e v: {angle:.1f} graus")
print(f"\nEm um neuronio: z = w . x + b = soma ponderada das features")
````

## 4. Produto Interno (Dot Product) <a id='4-produto-interno'></a>

### Intuicao: A Operacao Mais Importante de ML

O produto interno responde a uma pergunta fundamental: **quanto dois vetores "concordam" em direcao?**

Pense em duas pessoas avaliando filmes de 1 a 5. Se ambas dao notas altas aos mesmos filmes e baixas aos mesmos filmes, o produto interno sera grande (gostos similares). Se uma adora o que a outra odeia, sera negativo (gostos opostos).

### Definicao Formal

$$\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^{n} u_i v_i = u_1 v_1 + u_2 v_2 + \cdots + u_n v_n$$

**Relacao com angulo (vetores unitarios):**

$$\mathbf{u} \cdot \mathbf{v} = \|\mathbf{u}\| \|\mathbf{v}\| \cos\theta$$

| Angulo $\theta$ | $\cos\theta$ | Interpretacao |
|-----------------|--------------|---------------|
| 0 graus | +1 | Vetores identicos em direcao |
| 0-90 graus | >0 | Vetores "concordam" parcialmente |
| 90 graus | 0 | Vetores ortogonais (independentes) |
| 90-180 graus | <0 | Vetores "discordam" |
| 180 graus | -1 | Vetores opostos |

### Onde aparece em ML?

- **Neuronio artificial:** $z = \mathbf{w} \cdot \mathbf{x} + b$ (produto interno entre pesos e features)
- **Atencao (Transformers):** $\text{score} = \mathbf{q} \cdot \mathbf{k} / \sqrt{d}$
- **SVM:** kernel linear = produto interno
- **Recomendacao:** score = vetor_usuario . vetor_item

### por que em ml
- Distância é o fundamento de algoritmos: KNN, K-means, clustering hierárquico
- Escolher a métrica certa é crucial para o desempenho (Euclidiana vs Manhattan)
### conexao com
Produto interno é o coração de: regressão linear, redes neurais, atenção em transformers (0_4, 0_7).


### O que observar nos graficos acima

**Produto interno vs angulo (esquerda):** A curva mostra que o produto interno de vetores unitarios e simplesmente o cosseno do angulo entre eles. Quando os vetores apontam na mesma direcao (0 graus), o produto e maximo (+1). A 90 graus (ortogonais), e zero. A 180 graus (opostos), e -1. A area verde marca onde os vetores "concordam" e a vermelha onde "discordam".

**Attention scores (direita):** Esta e uma versao simplificada do mecanismo de atencao usado em Transformers (GPT, BERT). Cada celula mostra o produto interno entre a query de um token e a key de outro, dividido por raiz(d). Valores maiores (mais claros) indicam que o modelo "presta mais atencao" naquela combinacao. Todo o mecanismo de atencao que alimenta LLMs modernos se reduz a esta operacao: produto interno entre vetores.

**O que concluir:** O produto interno e a operacao mais importante de algebra linear para ML. Ele mede alinhamento entre vetores, e aparece em neuronios artificiais (z = w.x + b), mecanismos de atencao, SVMs, e sistemas de recomendacao.

## 5. Norma de Vetores (L1, L2, Lp) <a id='5-normas'></a>

A **norma** mede o "comprimento" ou "magnitude" de um vetor.

### Família de normas Lp

$$\|\mathbf{v}\|_p = \left(\sum_{i=1}^{n} |v_i|^p\right)^{1/p}$$

| Norma | $p$ | Fórmula | Uso em ML |
|-------|-----|---------|-----------|
| **L1** | 1 | $\sum |v_i|$ | Regularização Lasso, MAE |
| **L2** | 2 | $\sqrt{\sum v_i^2}$ | Regularização Ridge, distância Euclidiana |
| **L∞** | ∞ | $\max |v_i|$ | Robustez a outliers |
| **L0** | 0 | Conta não-zeros | Esparsidade (NP-hard) |

### Intuição Geométrica: Bolas Unitárias

Cada norma define uma forma diferente para o conjunto de vetores com norma = 1:
- **L1:** Losango (diamante)
- **L2:** Círculo (esfera)
- **L∞:** Quadrado


### conexao com
Similaridade de cosseno é usada em processamento de linguagem natural, recuperação de informação (0_4)

### por que em ml
A escolha de norma é crítica: L1 regularização causa esparsidade (feature selection), L2 suaviza.


````python
# Visualizacao: bolas unitarias L1, L2, Linf
fig, axes = plt.subplots(1, 3, figsize=(14, 4.5))

theta = np.linspace(0, 2*np.pi, 500)

# --- L2: circulo ---
axes[0].plot(np.cos(theta), np.sin(theta), 'b-', lw=2.5)
axes[0].fill(np.cos(theta), np.sin(theta), alpha=0.1, color='blue')
axes[0].set_title('Bola Unitaria L2 (Ridge)\n||v||_2 = 1 --> circulo')
axes[0].set_aspect('equal'); axes[0].set_xlim(-1.5, 1.5); axes[0].set_ylim(-1.5, 1.5)
axes[0].axhline(0, color='k', lw=0.3); axes[0].axvline(0, color='k', lw=0.3)

# --- L1: losango ---
l1_x = [1, 0, -1, 0, 1]
l1_y = [0, 1, 0, -1, 0]
axes[1].plot(l1_x, l1_y, 'r-', lw=2.5)
axes[1].fill(l1_x, l1_y, alpha=0.1, color='red')
axes[1].set_title('Bola Unitaria L1 (Lasso)\n||v||_1 = 1 --> losango')
axes[1].set_aspect('equal'); axes[1].set_xlim(-1.5, 1.5); axes[1].set_ylim(-1.5, 1.5)
axes[1].axhline(0, color='k', lw=0.3); axes[1].axvline(0, color='k', lw=0.3)
# Os cantos do losango tocam os eixos: e aqui que Lasso zera coeficientes
axes[1].scatter([1, -1, 0, 0], [0, 0, 1, -1], color='red', s=60, zorder=5)
axes[1].annotate('Cantos nos eixos:\npesos = 0 (esparsidade!)',
                 xy=(1, 0), xytext=(0.3, -1.2),
                 arrowprops=dict(arrowstyle='->', color='red'), color='red', fontsize=9)

# --- L-infinito: quadrado ---
linf_x = [1, 1, -1, -1, 1]
linf_y = [1, -1, -1, 1, 1]
axes[2].plot(linf_x, linf_y, 'g-', lw=2.5)
axes[2].fill(linf_x, linf_y, alpha=0.1, color='green')
axes[2].set_title('Bola Unitaria L-inf\n||v||_inf = 1 --> quadrado')
axes[2].set_aspect('equal'); axes[2].set_xlim(-1.5, 1.5); axes[2].set_ylim(-1.5, 1.5)
axes[2].axhline(0, color='k', lw=0.3); axes[2].axvline(0, color='k', lw=0.3)

plt.tight_layout()
plt.show()

# --- Demonstracao numerica ---
v = np.array([3.0, -4.0])
print(f"v = {v}")
print(f"||v||_1   = |3| + |-4| = {np.sum(np.abs(v)):.1f}  (soma dos absolutos)")
print(f"||v||_2   = sqrt(9+16) = {np.linalg.norm(v):.1f}  (distancia Euclidiana)")
print(f"||v||_inf = max(|3|,|-4|) = {np.max(np.abs(v)):.1f}  (maior absoluto)")

print(f"\nEm regularizacao:")
print(f"  Ridge (L2): penaliza ||w||_2^2 --> pesos pequenos mas nao zero")
print(f"  Lasso (L1): penaliza ||w||_1   --> pesos exatamente zero (feature selection!)")
````

### O que observar nos graficos acima

**L2 -- circulo (esquerda):** A bola unitaria L2 e um circulo perfeito. Todos os pontos a distancia 1 da origem formam essa curva. Quando usamos regularizacao Ridge, estamos forçando os pesos do modelo a ficarem *dentro* desse circulo. O circulo nao tem cantos, entao os pesos encolhem uniformemente mas raramente chegam a zero exato.

**L1 -- losango (centro):** A bola unitaria L1 e um losango (diamante). Os cantos do losango tocam os eixos coordenados. Isso e crucial: quando a restricao de otimizacao "bate" num canto, um dos pesos vira exatamente zero. E por isso que Lasso faz **feature selection** automaticamente -- ele zera coeficientes de features irrelevantes.

**L-infinito -- quadrado (direita):** Aqui so importa o maior valor absoluto entre as componentes. Todos os vetores cujo maior componente tem modulo 1 estao na borda. Menos usado em ML classico, mas aparece em robustez adversarial (ataques L-inf perturbam todos os pixels por no maximo epsilon).

**O que concluir:** A geometria da norma define o "formato" da restricao. Ridge (L2) encolhe tudo. Lasso (L1) zera o que nao importa. Essa diferenca geometrica e a razao fundamental pela qual Lasso faz feature selection e Ridge nao.
### O que concluir
Cada norma tem propriedades geométricas e uso em ML: L1 para sparsidade, L2 para regularização suave.

## 6. Distância entre Vetores <a id='6-distancias'></a>

A distância entre dois vetores $\mathbf{u}$ e $\mathbf{v}$ é a norma da diferença:

$$d(\mathbf{u}, \mathbf{v}) = \|\mathbf{u} - \mathbf{v}\|$$

| Distância | Fórmula | Quando usar |
|-----------|---------|-------------|
| **Euclidiana** | $\sqrt{\sum(u_i - v_i)^2}$ | Dados contínuos, KNN, K-Means |
| **Manhattan** | $\sum |u_i - v_i|$ | Dados com outliers, grids |
| **Cosseno** | $1 - \cos(\theta)$ | Texto, embeddings (ignora magnitude) |
| **Mahalanobis** | $(\mathbf{u}-\mathbf{v})^T \Sigma^{-1}(\mathbf{u}-\mathbf{v})$ | Leva covariância em conta |

### Aplicação: K-Nearest Neighbors (KNN)

O KNN classifica um ponto pela classe mais frequente entre seus $k$ vizinhos mais próximos,
onde "mais próximo" é medido por distância (tipicamente Euclidiana).


### por que em ml
- Independência linear garante que features não são redundantes
- Multicolinearidade em regressão causa instabilidade: detecta com rank

### conexao com
Distâncias são fundamentais em algoritmos baseados em instância: KNN, K-means, clustering (0_4, 0_5).


````python
# Visualizacao: distancias e KNN
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

# --- Painel 1: KNN visual (ponto novo + k vizinhos) ---
np.random.seed(42)
n_samples = 40
X_demo = np.vstack([
    np.random.randn(n_samples//2, 2) + [1, 1],
    np.random.randn(n_samples//2, 2) + [4, 4]
])
y_demo = np.repeat([0, 1], n_samples//2)
new_point = np.array([0.5, 1.0])
k = 5

# Calcular distancias
distances = np.sqrt(np.sum((X_demo - new_point)**2, axis=1))
k_nearest_indices = np.argsort(distances)[:k]

# Plot
for label in [0, 1]:
    axes[0].scatter(X_demo[y_demo==label, 0], X_demo[y_demo==label, 1],
                   s=80, c=['steelblue', 'coral'][label], alpha=0.7, label=f'Classe {label}')
axes[0].scatter(new_point[0], new_point[1], s=200, marker='*', c='yellow',
               edgecolor='black', lw=2, label='Ponto novo', zorder=10)
for idx in k_nearest_indices:
    axes[0].plot([new_point[0], X_demo[idx, 0]], [new_point[1], X_demo[idx, 1]],
                'k--', alpha=0.3, lw=1)
axes[0].scatter(X_demo[k_nearest_indices, 0], X_demo[k_nearest_indices, 1],
               s=150, facecolor='none', edgecolor='red', lw=2.5, label=f'{k} vizinhos')
axes[0].set_title(f'KNN com k={k}: encontrando vizinhos mais proximos')
axes[0].legend(fontsize=9)
axes[0].set_xlim(-2, 6); axes[0].set_ylim(-2, 6)

# --- Painel 2: Distribuição de distâncias ---
axes[1].hist(distances, bins=15, color='steelblue', alpha=0.7, edgecolor='black')
axes[1].axvline(distances[k_nearest_indices[-1]], color='red', linestyle='--', lw=2,
               label=f'Distância do {k}-ésimo vizinho')
axes[1].set_xlabel('Distância Euclidiana'); axes[1].set_ylabel('Frequência')
axes[1].set_title('Distribuição de distâncias do novo ponto')
axes[1].legend()

plt.tight_layout()
plt.show()
````

### O que observar nos graficos acima

**KNN visual (esquerda):** A estrela preta e um novo ponto que queremos classificar. O algoritmo calcula a distancia Euclidiana para *todos* os pontos de treinamento, seleciona os k=5 mais proximos (circulos com borda preta), e "vota" pela classe mais frequente entre eles. As linhas tracejadas mostram as distancias -- note que o ponto e classificado pela vizinhanca local, nao por uma regra global.

**Decision boundary (direita):** O KNN cria um "mapa" de decisao ao classificar cada ponto do espaco. A fronteira entre as regioes azul e vermelha e o decision boundary. Com k=5, o boundary e razoavelmente suave. Com k=1 seria extremamente irregular (overfitting: cada ponto de treino cria sua "ilha"). Com k=40 seria quase linear (underfitting: ignora a estrutura local).

**O que concluir:** KNN e conceitualmente simples -- classifica por proximidade. Mas traz consequencias importantes: a distancia Euclidiana assume que todas as features tem a mesma importancia. Se uma feature tem valores na casa dos milhares e outra entre 0 e 1, a primeira vai dominar completamente o calculo de distancia. Por isso, normalizar os dados antes de KNN e absolutamente essencial.

## 7. Ângulos e Similaridade de Cosseno <a id='7-cosseno'></a>

A **similaridade de cosseno** mede o ângulo entre dois vetores, independente de sua magnitude:

$$\text{sim}(\mathbf{u}, \mathbf{v}) = \cos\theta = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|}$$

Varia de $-1$ (opostos) a $+1$ (idênticos), com $0$ indicando ortogonalidade.

### Quando usar Cosseno vs. Euclidiana?

| Cenário | Métrica preferida |
|---------|-------------------|
| Texto: "Python é ótimo" vs "Python é incrível Python" | **Cosseno** (ignora frequência/magnitude) |
| Coordenadas geográficas (lat/lon) | **Euclidiana** (magnitude importa) |
| Embeddings de palavras (Word2Vec, BERT) | **Cosseno** |
| Anomalia em sensores (valor absoluto importa) | **Euclidiana** |
| Recomendação de filmes (ratings normalizados) | **Cosseno** |


### O que observar
- Quando projetamos dados em direções principais, reduzimos dimensionalidade sem perder muita informação
- PCA encontra as direções (autovetores) que maximizam variância

### por que em ml
Cosseno é usado em processamento de linguagem natural (busca textual, similaridade de documentos).


````python
# Visualizacao: similaridade de cosseno
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Painel 1: Matriz de similaridade entre 'documentos' ---
# Simulando vetores TF-IDF simplificados para 6 'documentos'
doc_names = ['ML paper', 'DL tutorial', 'Python guide', 'Receita bolo', 'Receita pizza', 'Math book']
docs = np.array([
    [3, 1, 1, 0, 0, 0],  # ML paper: machine, learning, python
    [2, 2, 1, 0, 0, 0],  # DL tutorial: deep, learning, python
    [0, 0, 3, 0, 0, 0],  # Python guide
    [0, 0, 0, 3, 1, 0],  # Receita bolo
    [0, 0, 0, 1, 3, 0],  # Receita pizza
    [0, 0, 0, 0, 0, 3],  # Math book
])

# Calcular matriz de similaridade de cosseno
def cos_sim(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v) + 1e-10)

n_docs = len(docs)
similarity_matrix = np.zeros((n_docs, n_docs))
for i in range(n_docs):
    for j in range(n_docs):
        similarity_matrix[i, j] = cos_sim(docs[i], docs[j])

im1 = axes[0].imshow(similarity_matrix, cmap='RdYlGn', vmin=0, vmax=1, aspect='auto')
axes[0].set_xticks(range(n_docs)); axes[0].set_yticks(range(n_docs))
axes[0].set_xticklabels(doc_names, rotation=45, ha='right', fontsize=8)
axes[0].set_yticklabels(doc_names, fontsize=8)
axes[0].set_title('Matriz de Similaridade de Cosseno')
plt.colorbar(im1, ax=axes[0], label='Cosseno')

# --- Painel 2: Query vs documentos ---
query = np.array([2, 1, 1, 0, 0, 0])  # "machine learning python"
query_sims = np.array([cos_sim(query, doc) for doc in docs])
colors_bar = ['green' if s > 0.5 else 'orange' if s > 0.1 else 'red' for s in query_sims]
axes[1].barh(range(n_docs), query_sims, color=colors_bar, alpha=0.8)
axes[1].set_yticks(range(n_docs)); axes[1].set_yticklabels(doc_names)
axes[1].set_xlabel('Similaridade de Cosseno')
axes[1].set_title('Ranking de Relevância (query: "machine learning python")')
for i, v in enumerate(query_sims):
    axes[1].text(v + 0.02, i, f'{v:.3f}', va='center', fontsize=9)

plt.tight_layout()
plt.show()
````

### O que observar nos graficos acima

**Matriz de similaridade (esquerda):** Os 6 "documentos" formam dois blocos claros. Documentos de tecnologia (ML paper, DL tutorial, Python guide) tem alta similaridade entre si (valores altos, amarelados). Documentos de culinaria (Receita bolo, Receita pizza) tambem sao similares entre si. Mas entre os blocos, a similaridade e quase zero -- como esperado, porque nao compartilham vocabulario. "Dieta fitness" tem similaridade intermediaria com culinaria (compartilha "ingrediente") mas zero com tech.

**Cosseno vs Euclidiana (direita):** Este e o ponto mais importante. O "Doc curto" e o "Doc longo" apontam na mesma direcao (mesmo topico, mas um e mais extenso). A Euclidiana diz que eles estao longe (4.47) porque um vetor e 3x maior. O Cosseno diz que sao identicos (1.00) porque a direcao e a mesma. Ja o "Doc diferente" esta Euclidianamente perto (3.16) do Doc curto, mas o Cosseno revela que sao sobre topicos completamente diferentes (0.00).

**O que concluir:** Para texto e embeddings, cosseno e quase sempre a metrica correta. Euclidiana penaliza documentos longos injustamente. Em sistemas de busca (Google, RAGs), recomendacao (Netflix), e NLP em geral, cosseno e o padrao. Euclidiana e melhor quando a magnitude realmente importa (ex: coordenadas fisicas, valores de sensores).

## 8. Vetores Unitários e Normalização <a id='8-normalizacao'></a>

Um **vetor unitário** tem norma 1: $\|\hat{\mathbf{v}}\| = 1$

Para criar um vetor unitário a partir de qualquer vetor não-nulo:

$$\hat{\mathbf{v}} = \frac{\mathbf{v}}{\|\mathbf{v}\|}$$

### Por que normalizar em ML?

1. **Algoritmos baseados em distância:** KNN, K-Means, SVM ficam sensíveis à escala
   sem normalização → uma feature com valores maiores domina
2. **Gradient descent:** converge muito mais rápido com features normalizadas
3. **Similaridade de cosseno:** requer vetores para ser interpretável
4. **Redes neurais:** BatchNorm, LayerNorm — normalização é onipresente
### conexao com
Normalização é obrigatória para algoritmos sensíveis a escala: KNN, SVM, redes neurais (0_4, 0_5).


````python
# Visualizacao: efeito da normalizacao em dados reais
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Gerar dados com escalas muito diferentes
np.random.seed(42)
n = 60
altura = np.random.normal(1.70, 0.08, n)       # metros: ~1.55 a ~1.85
renda = np.random.normal(3000, 800, n)          # reais: 1500 a 4500

X_raw = np.column_stack([altura, renda])

# Normalizacao manual: (X - mean) / std
X_normalized = (X_raw - X_raw.mean(axis=0)) / X_raw.std(axis=0)

# Plot antes vs depois
axes[0].scatter(X_raw[:, 0], X_raw[:, 1], s=80, alpha=0.6, color='coral')
axes[0].set_xlabel('Altura (metros)'); axes[0].set_ylabel('Renda (R$)')
axes[0].set_title('Dados Brutos (escalas DIFERENTES)')
axes[0].annotate(f'Altura: range {X_raw[:, 0].min():.2f}-{X_raw[:, 0].max():.2f}',
                xy=(0.5, 0.05), xycoords='axes fraction', fontsize=9)
axes[0].annotate(f'Renda: range {X_raw[:, 1].min():.0f}-{X_raw[:, 1].max():.0f}',
                xy=(0.5, 0.01), xycoords='axes fraction', fontsize=9)

axes[1].scatter(X_normalized[:, 0], X_normalized[:, 1], s=80, alpha=0.6, color='steelblue')
axes[1].set_xlabel('Altura (normalizada)'); axes[1].set_ylabel('Renda (normalizada)')
axes[1].set_title('Dados Normalizados (escalas IGUAIS)')
axes[1].axhline(0, color='k', lw=0.5); axes[1].axvline(0, color='k', lw=0.5)
axes[1].grid(True, alpha=0.3)

print(f"Escalas antes:  altura={X_raw[:, 0].std():.3f}, renda={X_raw[:, 1].std():.0f}")
print(f"Escalas depois: altura={X_normalized[:, 0].std():.3f}, renda={X_normalized[:, 1].std():.3f}")
print(f"\nCom escalas diferentes, KNN favorecia 'renda' (maior escala).")
print(f"Com normalizacao, ambas as features tem peso igual!")
````

### O que observar nos graficos acima

**Sem normalizacao (esquerda):** Os dados estao esticados horizontalmente porque a renda varia de ~1400 a ~4600 (range de ~3200), enquanto a altura varia de ~1.55 a ~1.85 (range de ~0.30). A distancia Euclidiana entre dois pontos sera quase inteiramente determinada pela renda. A altura, mesmo sendo muito diferente, contribui com centesimos de unidade para a distancia total. Resultado: KNN ignoraria completamente a altura.

**Com normalizacao (direita):** Apos aplicar StandardScaler (subtrai a media, divide pelo desvio padrao), ambas as features tem media 0 e variancia 1. Agora a distribuicao dos pontos e equilibrada e a distancia Euclidiana leva ambas as features igualmente em conta. Note que o aspecto do grafico e 'equal' -- os eixos tem a mesma escala.

**Exemplo numerico:** P1 e P2 diferem muito em renda mas nao em altura. P1 e P3 diferem muito em altura mas nao em renda. Sem normalizar, dist(P1,P2) = 500 e dist(P1,P3) = 10. O KNN concluiria que P3 esta "muito mais perto" que P2, quando na verdade ambos diferem em uma feature.

**O que concluir:** Sempre normalize antes de usar algoritmos baseados em distancia (KNN, K-Means, SVM). Os dois metodos mais comuns sao StandardScaler (z-score: media=0, std=1) e MinMaxScaler (mapeia para [0,1]). Em redes neurais, BatchNorm e LayerNorm fazem normalizacao internamente entre camadas.

## 9. Espaços Vetoriais (Conceitos) <a id='9-espacos-vetoriais'></a>

Um **espaço vetorial** é um conjunto $V$ com operações de adição e multiplicação por escalar
que satisfazem certas axiomas (fechamento, associatividade, etc.).

### Conceitos-chave para ML

**Combinação Linear:**
$$\mathbf{w} = c_1 \mathbf{v}_1 + c_2 \mathbf{v}_2 + \cdots + c_k \mathbf{v}_k$$

**Base:** conjunto mínimo de vetores que gera todo o espaço (lineamente independentes).

**Span:** conjunto de todas as combinações lineares de um grupo de vetores.

**Independência Linear:** $\{\mathbf{v}_1, \ldots, \mathbf{v}_k\}$ são L.I. se
$c_1\mathbf{v}_1 + \cdots + c_k\mathbf{v}_k = \mathbf{0}$ implica todos $c_i = 0$.

### Por que importa em ML?

- **Rank da matriz de features:** determina se o problema de regressão tem solução única
- **Espaço nulo:** features redundantes ficam no espaço nulo (correlação perfeita)
- **Embeddings:** representações em subespaços de menor dimensão
- **Atenção:** operações em subespaços das queries, keys e values
### por que em ml
Espaços vetoriais definem formalmente onde nossos dados vivem e garantem que operações sejam bem-definidas.


````python
# Visualizacao: espacos vetoriais, base e independencia linear
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# --- Painel 1: Base canonica e combinacoes lineares ---
# Base canonica: e1=[1,0], e2=[0,1]
e1 = np.array([1, 0])
e2 = np.array([0, 1])

axes[0].annotate('', xy=e1*1.5, xytext=(0,0),
                arrowprops=dict(arrowstyle='->', color='blue', lw=2.5))
axes[0].annotate('', xy=e2*1.5, xytext=(0,0),
                arrowprops=dict(arrowstyle='->', color='red', lw=2.5))
axes[0].text(1.6, -0.2, 'e1', color='blue', fontsize=12, fontweight='bold')
axes[0].text(-0.3, 1.6, 'e2', color='red', fontsize=12, fontweight='bold')

# Combinacoes lineares: v = a*e1 + b*e2
combos = [(2, 1, '#4CAF50'), (1, 3, '#FF9800'), (-1, 2, '#9C27B0'), (3, -1, '#795548')]
for a_coeff, b_coeff, c in combos:
    v = a_coeff * e1 + b_coeff * e2
    axes[0].annotate('', xy=v, xytext=(0,0),
                    arrowprops=dict(arrowstyle='->', color=c, lw=1.5, ls='--'))
    axes[0].plot(*v, 'o', color=c, ms=6)
    axes[0].text(v[0]+0.1, v[1]+0.1, f'{a_coeff}e1+{b_coeff}e2', fontsize=8, color=c)

axes[0].set_xlim(-2, 4); axes[0].set_ylim(-2, 4)
axes[0].axhline(0, color='k', lw=0.3); axes[0].axvline(0, color='k', lw=0.3)
axes[0].set_aspect('equal')
axes[0].set_title('Vetores L.I. (base canonica)\nSpan = todo R^2')

# --- Painel 2: Vetores linearmente DEPENDENTES ---
va = np.array([1, 2])
vb = np.array([2, 4])  # = 2 * va (dependente!)

# Span: apenas uma reta
t = np.linspace(-2, 2, 50)
line_x = t * va[0]
line_y = t * va[1]
axes[1].plot(line_x, line_y, 'gray', ls='--', lw=1, alpha=0.5)

axes[1].annotate('', xy=va, xytext=(0,0),
                arrowprops=dict(arrowstyle='->', color='blue', lw=2.5))
axes[1].annotate('', xy=vb, xytext=(0,0),
                arrowprops=dict(arrowstyle='->', color='red', lw=2.5))
axes[1].text(va[0]+0.1, va[1]+0.1, 'va=[1,2]', color='blue', fontsize=10, fontweight='bold')
axes[1].text(vb[0]+0.1, vb[1]+0.1, 'vb=[2,4]=2*va', color='red', fontsize=10, fontweight='bold')

axes[1].set_xlim(-3, 4); axes[1].set_ylim(-3, 5)
axes[1].axhline(0, color='k', lw=0.3); axes[1].axvline(0, color='k', lw=0.3)
axes[1].set_aspect('equal')
axes[1].set_title('Vetores L.D. (dependentes)\nSpan = apenas uma reta!')
axes[1].text(-2.5, -2, 'vb = 2*va\nRedundancia!\nSo cobrem 1 direcao', fontsize=9,
            bbox=dict(facecolor='lightyellow', alpha=0.8))

# --- Painel 3: Analogia com features em ML ---
# Exemplo: feature1 = idade, feature2 = 2*idade (redundante)
np.random.seed(42)
idade = np.random.uniform(20, 60, 30)
idade_dobro = 2 * idade + np.random.normal(0, 0.5, 30)  # quase perfeitamente correlacionado
renda_real = np.random.uniform(2000, 8000, 30)

axes[2].scatter(idade, idade_dobro, c='red', s=30, alpha=0.7, label='idade vs 2*idade\n(L.D. / redundante)')
axes[2].scatter(idade[:15], renda_real[:15], c='blue', s=30, alpha=0.7, label='idade vs renda\n(L.I. / util)')

axes[2].set_xlabel('Feature 1')
axes[2].set_ylabel('Feature 2')
axes[2].set_title('Features em ML:\nL.D. = redundancia, L.I. = informacao nova')
axes[2].legend(fontsize=9, loc='upper left')

plt.tight_layout()
plt.show()

# Demonstracao numerica
print("=== Independencia Linear ===")
v1 = np.array([1, 0])
v2 = np.array([0, 1])
v3 = np.array([2, 3])  # = 2*v1 + 3*v2
print(f"v1={v1}, v2={v2}: L.I. (rank=2, cobrem R^2)")
print(f"v3={v3} = 2*v1 + 3*v2: v3 e redundante (nao adiciona informacao)")
print(f"\nEm ML: se feature3 = 2*feature1 + 3*feature2,")
print(f"a matriz de features tera rank < 3 (colinearidade).")
````

### O que observar nos graficos acima

**Base canonica e combinacoes lineares (esquerda):** Os vetores e1=[1,0] e e2=[0,1] sao perpendiculares e formam uma base para R^2. Qualquer vetor no plano pode ser escrito como a*e1 + b*e2 (as setas tracejadas). Os coeficientes a e b sao simplesmente as coordenadas do vetor. Isso e exatamente o que one-hot encoding faz: cada categoria e representada por um vetor da base canonica.

**Vetores dependentes (centro):** va=[1,2] e vb=[2,4] apontam na mesma direcao (vb = 2*va). Mesmo tendo dois vetores, so conseguimos gerar pontos ao longo de uma unica reta (a reta cinza tracejada). Ter vb alem de va nao adiciona informacao nenhuma -- e pura redundancia. O "span" e uma reta, nao o plano inteiro.

**Analogia em ML (direita):** Os pontos vermelhos mostram duas features que sao quase perfeitamente correlacionadas (idade e 2*idade). Eles formam uma reta, assim como os vetores dependentes. Os pontos azuis mostram features independentes (idade e renda) que se espalham pelo plano. Features independentes trazem informacao nova; features dependentes sao desperdiço de dimensionalidade.

**O que concluir:** Colinearidade (dependencia linear) entre features e um problema real em ML. Regressao linear pode ter infinitas solucoes, a inversa da matriz nao existe, e os coeficientes ficam instaveis. PCA resolve isso encontrando as direcoes verdadeiramente independentes e descartando as redundantes.

## 10. Projeção de Vetores → PCA (Intuição) <a id='10-projecao'></a>

A **projeção** de $\mathbf{u}$ sobre $\mathbf{v}$ é o vetor que representa
"quanto de $\mathbf{u}$ está na direção de $\mathbf{v}$":

$$\text{proj}_{\mathbf{v}} \mathbf{u} = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{v}\|^2} \mathbf{v} = \frac{\mathbf{u} \cdot \hat{\mathbf{v}}}{1} \hat{\mathbf{v}}$$

### Conexão com PCA

O **PCA (Principal Component Analysis)** usa projeções para encontrar as direções
(componentes principais) que capturam a **maior variância** dos dados.

1. Calcula a direção de maior variância (1º componente principal $\mathbf{v}_1$)
2. **Projeta** todos os dados nessa direção: $z_i = \mathbf{x}_i \cdot \mathbf{v}_1$
3. Remove essa componente e repete para o 2º componente

> Isso é exatamente o que você vai aprender em detalhes no Notebook 0.3
> (autovalores/autovetores) e no Notebook 3.6 (Redução de Dimensionalidade).
### O que concluir
Projeção reduz dimensionalidade sem perder informação: é a essência de PCA e análise de componentes principais.


````python
# Visualizacao: projecao vetorial e PCA
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Gerar dados correlacionados 2D
np.random.seed(42)
n_pts = 80
mean = [2, 3]
cov = [[2, 1.8], [1.8, 2]]  # alta correlacao
data = np.random.multivariate_normal(mean, cov, n_pts)

# Calcular PCA manualmente (para fins didaticos)
data_centered = data - data.mean(axis=0)
cov_matrix = np.cov(data_centered.T)
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)
# Ordenar do maior para o menor autovalor
idx = eigenvalues.argsort()[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]
pc1 = eigenvectors[:, 0]  # direcao de maior variancia
pc2 = eigenvectors[:, 1]  # direcao perpendicular

# --- Painel 1: Dados originais com componentes principais ---
axes[0].scatter(data[:, 0], data[:, 1], c='steelblue', s=25, alpha=0.6)
# Plotar PC1 e PC2 como setas (escala proporcional ao autovalor)
origin = data.mean(axis=0)
scale1 = np.sqrt(eigenvalues[0]) * 2
scale2 = np.sqrt(eigenvalues[1]) * 2
axes[0].annotate('', xy=origin + pc1*scale1, xytext=origin,
                arrowprops=dict(arrowstyle='->', color='red', lw=3))
axes[0].annotate('', xy=origin + pc2*scale2, xytext=origin,
                arrowprops=dict(arrowstyle='->', color='orange', lw=3))
axes[0].text(origin[0] + pc1[0]*scale1 + 0.1, origin[1] + pc1[1]*scale1 + 0.1,
            f'PC1 ({eigenvalues[0]/eigenvalues.sum()*100:.0f}% var)', color='red', fontsize=10, fontweight='bold')
axes[0].text(origin[0] + pc2[0]*scale2 - 1.2, origin[1] + pc2[1]*scale2 + 0.1,
            f'PC2 ({eigenvalues[1]/eigenvalues.sum()*100:.0f}% var)', color='orange', fontsize=10, fontweight='bold')
axes[0].set_title('Dados 2D com Componentes Principais')
axes[0].set_xlabel('Feature 1'); axes[0].set_ylabel('Feature 2')
axes[0].set_aspect('equal')

# --- Painel 2: Projecao sobre PC1 ---
# Projetar cada ponto na direcao de PC1
projections = data_centered @ pc1  # coordenada escalar ao longo de PC1
proj_points = np.outer(projections, pc1) + data.mean(axis=0)  # pontos na reta de PC1

axes[1].scatter(data[:, 0], data[:, 1], c='steelblue', s=25, alpha=0.4)
# Reta de PC1
t = np.linspace(-4, 4, 100)
reta_x = origin[0] + t * pc1[0]
reta_y = origin[1] + t * pc1[1]
axes[1].plot(reta_x, reta_y, 'r-', lw=2, label='PC1')
# Linhas de projecao (erro de reconstrucao)
for i in range(0, n_pts, 3):  # cada 3 para nao sobrecarregar
    axes[1].plot([data[i, 0], proj_points[i, 0]], [data[i, 1], proj_points[i, 1]],
                'gray', lw=0.5, alpha=0.5)
axes[1].scatter(proj_points[:, 0], proj_points[:, 1], c='red', s=15, alpha=0.7)
axes[1].set_title('Projecao sobre PC1\n(linhas cinzas = erro de reconstrucao)')
axes[1].set_xlabel('Feature 1'); axes[1].set_ylabel('Feature 2')
axes[1].set_aspect('equal')
axes[1].legend()

# --- Painel 3: Dados no espaco PCA ---
data_pca = data_centered @ eigenvectors  # transformar para espaco PCA

axes[2].scatter(data_pca[:, 0], data_pca[:, 1], c='steelblue', s=25, alpha=0.6)
axes[2].axhline(0, color='k', lw=0.5); axes[2].axvline(0, color='k', lw=0.5)
axes[2].set_xlabel(f'PC1 ({eigenvalues[0]/eigenvalues.sum()*100:.0f}% variancia)')
axes[2].set_ylabel(f'PC2 ({eigenvalues[1]/eigenvalues.sum()*100:.0f}% variancia)')
axes[2].set_title('Dados no espaco PCA\n(componentes agora sao nao-correlacionados)')
# Indicar que PC2 poderia ser descartado
axes[2].annotate('PC2 tem pouca variancia\n--> pode ser descartado!',
                xy=(0, data_pca[:, 1].max()), xytext=(1.5, data_pca[:, 1].max()),
                arrowprops=dict(arrowstyle='->', color='orange'),
                color='orange', fontsize=9)

plt.tight_layout()
plt.show()

# Demonstracao numerica da projecao
print("=== Projecao: quanto de u esta na direcao de v? ===")
u = np.array([3.0, 4.0])
v = np.array([1.0, 0.0])  # eixo x
proj_scalar = np.dot(u, v) / np.linalg.norm(v)
proj_vec = (np.dot(u, v) / np.dot(v, v)) * v
print(f"u = {u}, v = {v}")
print(f"Projecao escalar de u sobre v = {proj_scalar:.1f}")
print(f"Projecao vetorial = {proj_vec}")
print(f"\nInterpretacao: dos 5 'unidades' de comprimento de u,")
print(f"3 estao na direcao horizontal (v) e 4 na vertical (perpendicular a v).")
````

### O que observar nos graficos acima

**Dados com componentes principais (esquerda):** Os dados 2D tem uma correlacao forte (se espalham ao longo de uma diagonal). PC1 (seta vermelha) aponta na direcao de maior variacao -- a diagonal. PC2 (seta laranja) aponta na direcao perpendicular, onde ha pouca variacao. O comprimento das setas e proporcional a variancia capturada: PC1 e muito maior que PC2.

**Projecao sobre PC1 (centro):** Todos os pontos sao "projetados" na reta vermelha (PC1). Cada ponto azul se torna um ponto vermelho na reta. As linhas cinzas mostram o erro de reconstrucao -- a informacao que perdemos ao reduzir de 2D para 1D. Note que as linhas sao curtas: perdemos pouco ao descartar PC2.

**Espaco PCA (direita):** Apos a transformacao, os dados estao em um novo sistema de coordenadas onde PC1 e PC2 sao os eixos. Agora as duas componentes sao nao-correlacionadas (nao ha mais o padrao diagonal). A variancia esta concentrada em PC1, e PC2 contribui pouco. Poderiamos descartar PC2 e manter ~95% da informacao.

**O que concluir:** PCA e essencialmente projecao vetorial repetida. Encontra as direcoes de maior variancia, projeta os dados nelas, e descarta as direcoes com pouca variancia. Em problemas reais (imagens com milhares de pixels), PCA pode reduzir centenas de dimensoes mantendo a maior parte da informacao. Isso acelera o treinamento e combate overfitting.

````python
# === SOLUCAO: Exercicio 3 ===

def knn_predict(X_train, y_train, x_test, k=3):
    # Passo 1: distancias Euclidianas
    distances = np.linalg.norm(X_train - x_test, axis=1)
    
    # Passo 2: indices dos k menores
    k_nearest_idx = np.argsort(distances)[:k]
    
    # Passo 3: labels dos vizinhos
    k_labels = y_train[k_nearest_idx]
    
    # Passo 4: retorna a moda (label mais frequente)
    values, counts = np.unique(k_labels, return_counts=True)
    return values[np.argmax(counts)]

# Dados de treinamento
np.random.seed(42)
X_train = np.vstack([
    np.random.randn(25, 2) + [1, 1],
    np.random.randn(25, 2) + [4, 4]
])
y_train = np.repeat([0, 1], 25)

# Classifique
x_test1 = np.array([1.5, 1.5])
x_test2 = np.array([4.0, 4.5])

pred1 = knn_predict(X_train, y_train, x_test1, k=5)
pred2 = knn_predict(X_train, y_train, x_test2, k=5)

print(f"x_test1 = {x_test1} -> predição: classe {int(pred1)} (esperado: classe 0)")
print(f"x_test2 = {x_test2} -> predição: classe {int(pred2)} (esperado: classe 1)")

# Visualizar
fig, ax = plt.subplots(figsize=(10, 8))
ax.scatter(X_train[y_train==0, 0], X_train[y_train==0, 1], s=80, c='steelblue', alpha=0.7, label='Classe 0')
ax.scatter(X_train[y_train==1, 0], X_train[y_train==1, 1], s=80, c='coral', alpha=0.7, label='Classe 1')
ax.scatter([x_test1[0], x_test2[0]], [x_test1[1], x_test2[1]], s=200, marker='*',
          c=['green', 'purple'], edgecolor='black', lw=2, label='Testes', zorder=10)
ax.set_xlabel('Feature 1'); ax.set_ylabel('Feature 2')
ax.set_title('KNN com k=5: Classificação de Novos Pontos')
ax.legend()
plt.tight_layout()
plt.show()
````

## 12. Erros Comuns e Armadilhas <a id='12-erros'></a>

### Erro 1: Nao normalizar antes de KNN/K-Means

**Sintoma:** Features com valores grandes dominam completamente a distancia, ignorando features com valores pequenos.

**Exemplo:** Renda (0-100000) vs Idade (0-80). A renda contribui com ~99.99% da distancia.

**Solucao:** Sempre usar StandardScaler ou MinMaxScaler antes de algoritmos baseados em distancia.

```python
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(X)
```

---

### Erro 2: Confundir similaridade de cosseno com distancia de cosseno

**Armadilha:** Similaridade de cosseno varia de -1 a +1 (quanto MAIOR, mais similar). Distancia de cosseno = 1 - similaridade (quanto MENOR, mais proximo). Misturar os dois inverte o ranking.

**Regra:** Em busca/recomendacao, use `similaridade` e ordene do maior para menor. Em clustering (K-Means), use `distancia` e ordene do menor para maior.

---

### Erro 3: Usar Euclidiana para dados de alta dimensionalidade

**Sintoma:** Em dimensoes muito altas (>100), todos os pontos ficam "igualmente longe" uns dos outros (curse of dimensionality). A Euclidiana perde poder discriminativo.

**Solucao:** Para dados de alta dimensao (texto, embeddings), preferir cosseno. Ou reduzir dimensionalidade com PCA antes de calcular distancias.

---

### Erro 4: Confundir vetor (1,) com vetor (1,n) com matriz (1,n)

**Armadilha em NumPy:**
```python
a = np.array([1, 2, 3])        # shape (3,)  -- vetor 1D
b = np.array([[1, 2, 3]])      # shape (1,3) -- matrix 2D com 1 linha
c = np.array([[1], [2], [3]])   # shape (3,1) -- matrix 2D com 1 coluna

# a.T == a  (transposta de vetor 1D nao muda nada!)
# b.T.shape == (3,1)  (transposta funciona)
```

**Regra:** Se precisa de transposta, use arrays 2D. Vetores 1D em NumPy nao tem "orientacao" (nem linha nem coluna).

---

### Erro 5: Esquecer que produto interno requer mesma dimensao

**Erro:** `np.dot(np.array([1,2]), np.array([1,2,3]))` gera erro.

**Regra:** Vetores devem ter exatamente o mesmo numero de componentes para produto interno, soma, e distancia.### conexao com

## Conexão com ML
